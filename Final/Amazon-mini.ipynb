# Phase 2: Advanced Data Cleaning (Solving the "Messy Data" Challenge)

```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv('amazon_india_2015.csv')

# 1. Clean Price/Currency Columns
def clean_currency(df, column):
    return pd.to_numeric(df[column].astype(str).str.replace('₹', '', regex=False).str.replace(',', '', regex=False), errors='coerce')

currency_cols = ['original_price_inr', 'discounted_price_inr', 'subtotal_inr', 'final_amount_inr', 'delivery_charges']
for col in currency_cols:
    df[col] = clean_currency(df, col)

# 2. Fix Mixed Date Formats (The 3,252 missed rows fix)
df['order_date'] = pd.to_datetime(df['order_date'], format='mixed', dayfirst=True)

# 3. Standardize Boolean Columns
bool_map = {'true': True, 'false': False, 'yes': True, 'no': False, '1': True, '0': False}
cols_to_fix = ['is_prime_member', 'is_festival_sale', 'is_prime_eligible']
for col in cols_to_fix:
    df[col] = df[col].astype(str).str.strip().str.lower().map(bool_map).fillna(False).astype(bool)

# 4. Handle Missing Values & Rating Format (e.g., "4/5")
df['delivery_charges'] = df['delivery_charges'].fillna(0).astype(int)
df['customer_age_group'] = df['customer_age_group'].fillna('Unknown')
df['festival_name'] = df['festival_name'].fillna('Regular Day')

# Fix ratings: Convert "4/5" to 4.0 and handle NaNs
df['customer_rating'] = pd.to_numeric(df['customer_rating'].astype(str).str.split('/').str[0], errors='coerce')
df['customer_rating'] = df['customer_rating'].fillna(df['customer_rating'].median())

# 6. Clean Category Names
df['category'] = (
    df['category']
    .astype(str)
    .str.replace('&', 'and')   # Replace & with 'and'
    .str.strip()               # Remove extra spaces
)

# 5. Remove Duplicates
df = df.drop_duplicates(subset='transaction_id', keep='first')

# Final Verification
print(f"Data Cleaning Complete.")
print(f"Total Rows: {len(df)}")
print(f"Remaining null dates: {df['order_date'].isna().sum()}")
```

```python
# df['order_date'].iloc[61:100]
target_txn = 'TXN_2015_00000041'
result = df[df['transaction_id'] == target_txn][['transaction_id', 'order_date']]

print(result)
```

# Q3 ECA

RFM (Recency, Frequency, Monetary) analysis - powerful customer segmentation technique

Data Preparation

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Ensure date format
main_df['order_date'] = pd.to_datetime(main_df['order_date'])

# Create snapshot date (latest date + 1)
snapshot_date = main_df['order_date'].max() + pd.Timedelta(days=1)
```

# 2. RFM Calculation

```python
# RFM calculation
rfm = main_df.groupby('customer_id').agg(
    Recency=('order_date', lambda x: (snapshot_date - x.max()).days),
    Frequency=('transaction_id', 'nunique'),   
    Monetary=('final_amount_inr', 'sum')
).reset_index()

rfm.columns = ['Customer_ID', 'Recency', 'Frequency', 'Monetary']
```

# 3. RFM Scoring (1–5 scale)

```python
rfm['R_Score'] = pd.qcut(rfm['Recency'], 5, labels=[5,4,3,2,1])
rfm['F_Score'] = pd.qcut(rfm['Frequency'].rank(method='first'), 5, labels=[1,2,3,4,5])
rfm['M_Score'] = pd.qcut(rfm['Monetary'], 5, labels=[1,2,3,4,5])

rfm['RFM_Score'] = rfm['R_Score'].astype(str) + rfm['F_Score'].astype(str) + rfm['M_Score'].astype(str)
```

Customer Segmentation

```python
def segment(row):
    if row['RFM_Score'] in ['555','554','544','545','454','455']:
        return 'Champions'
    elif row['R_Score'] >= 4:
        return 'Recent Customers'
    elif row['F_Score'] >= 4:
        return 'Loyal Customers'
    elif row['M_Score'] >= 4:
        return 'Big Spenders'
    elif row['R_Score'] <= 2:
        return 'At Risk'
    else:
        return 'Others'

rfm['Segment'] = rfm.apply(segment, axis=1)
```

Build customer lifetime value (CLV) analysis using cohort analysis, retention curves, and CLV distribution across different customer segments and acquisition years.


Cohort Analysis: Retention Curves
Retention curves show the percentage of customers who continue to shop 1, 2, or 5 years after their first purchase. A "flatter" curve means high loyalty.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Identify First Purchase Year for each Customer
main_df['first_purchase_year'] = main_df.groupby('customer_id')['order_date'].transform(lambda x: x.min().year)
main_df['order_year'] = main_df['order_date'].dt.year
main_df['years_since_joining'] = main_df['order_year'] - main_df['first_purchase_year']

# 2. Calculate Retention (Count unique customers per cohort per year)
cohort_data = main_df.groupby(['first_purchase_year', 'years_since_joining'])['customer_id'].nunique().reset_index()
cohort_pivot = cohort_data.pivot(index='first_purchase_year', columns='years_since_joining', values='customer_id')

# 3. Convert to Percentages
retention_matrix = cohort_pivot.divide(cohort_pivot[0], axis=0)

plt.figure(figsize=(12, 8))
sns.heatmap(retention_matrix, annot=True, fmt=".0%", cmap="YlGnBu")
plt.title('Customer Retention Matrix: Cohorts by Acquisition Year', fontsize=15)
plt.show()
```

```python
# CLV Distribution (Histogram)
# Aggregate total spend per customer across the whole 10-year period
clv_base = main_df.groupby('customer_id').agg(
    total_spend=('final_amount_inr', 'sum'),
    total_orders=('transaction_id', 'count'),
    first_year=('first_purchase_year', 'first')
).reset_index()

plt.figure(figsize=(10, 6))
# Using a log scale because CLV usually varies from 500 to 5,00,000+
sns.histplot(clv_base['total_spend'], bins=50, kde=True, color='purple', log_scale=True)

plt.title('Distribution of Customer Lifetime Value (CLV)', fontsize=14)
plt.xlabel('Total Lifetime Spend (INR) - Log Scale')
plt.ylabel('Number of Customers')
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Aggregate Monthly Metrics
health_metrics = main_df.resample('M', on='order_date').agg(
    revenue=('final_amount_inr', 'sum'),
    new_customers=('customer_id', 'nunique'),
    avg_delivery_time=('delivery_days', 'mean')
).reset_index()

# 2. Calculate Retention Rate (Proxy: % of orders from repeat customers)
repeat_cust = main_df.groupby([main_df['order_date'].dt.to_period('M'), 'customer_id']).size().reset_index()
retention_stats = repeat_cust.groupby('order_date').apply(
    lambda x: (x.duplicated(subset=['customer_id']).sum() / len(x)) * 100
).reset_index(name='retention_rate')

# 3. Merge all into a single Health DataFrame
health_metrics['retention_rate'] = retention_stats['retention_rate']
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.tsa.holtwinters import ExponentialSmoothing

# 1. Sales Forecasting Preparation (Monthly Revenue)
forecast_data = main_df.resample('M', on='order_date')['final_amount_inr'].sum()

# 2. Customer Churn Identification
# Define 'Churn' as no purchase in the last 180 days from the max date
max_date = main_df['order_date'].max()
customer_recency = main_df.groupby('customer_id')['order_date'].max().reset_index()
customer_recency['days_since_last_purchase'] = (max_date - customer_recency['order_date']).dt.days
customer_recency['is_churned'] = (customer_recency['days_since_last_purchase'] > 180).astype(int)
```

```python
# Fit the model
model = ExponentialSmoothing(forecast_data, trend='add', seasonal='add', seasonal_periods=12).fit()
forecast = model.forecast(12)

plt.figure(figsize=(14, 7))
plt.plot(forecast_data.tail(36), label='Historical Sales (Last 3 Years)')
plt.plot(forecast, label='2026 Predictive Forecast', color='red', linestyle='--')

plt.title('Sales Forecasting: Predicted Revenue Trend for 2026', fontsize=15)
plt.ylabel('Revenue (INR)')
plt.fill_between(forecast.index, forecast*0.9, forecast*1.1, color='red', alpha=0.1, label='90% Confidence')
plt.legend()
plt.show()
```

```python
# Analyze churn by Prime Membership
churn_stats = main_df.merge(customer_recency[['customer_id', 'is_churned']], on='customer_id')
churn_summary = churn_stats.groupby('is_prime_member')['is_churned'].mean().reset_index()

plt.figure(figsize=(8, 6))
sns.barplot(data=churn_summary, x='is_prime_member', y='is_churned', palette='magma')

plt.title('Churn Risk: Prime vs. Non-Prime Customers', fontsize=14)
plt.ylabel('Churn Probability (%)')
plt.xlabel('Is Prime Member?')
plt.show()
```

Design a Market Intelligence Dashboard analyzing competitor tracking, market trends, pricing intelligence, and strategic positioning insights.

```python
# Data Preparation: Market Share & Pricing Power
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Calculate Market Share by Category
market_total = main_df.groupby('category')['final_amount_inr'].sum().reset_index()
market_total.columns = ['category', 'total_market_revenue']

brand_performance = main_df.groupby(['category', 'brand'])['final_amount_inr'].sum().reset_index()
brand_performance = brand_performance.merge(market_total, on='category')

# 2. Market Share %
brand_performance['market_share_pct'] = (brand_performance['final_amount_inr'] / brand_performance['total_market_revenue']) * 100

# 3. Pricing Intelligence: How expensive is a brand compared to the category average?
category_avg_price = main_df.groupby('category')['final_amount_inr'].mean().reset_index()
category_avg_price.columns = ['category', 'avg_cat_price']

brand_pricing = main_df.groupby(['category', 'brand'])['final_amount_inr'].mean().reset_index()
brand_pricing = brand_pricing.merge(category_avg_price, on='category')
brand_pricing['price_index'] = brand_pricing['final_amount_inr'] / brand_pricing['avg_cat_price']
```

```python
# A. Strategic Positioning: Market Share vs. Pricing Power
plt.figure(figsize=(12, 8))
# Filter for the top category to keep it clean
top_cat = brand_performance.sort_values('total_market_revenue', ascending=False)['category'].iloc[0]
plot_data = brand_pricing[brand_pricing['category'] == top_cat].merge(brand_performance, on=['category', 'brand'])

sns.scatterplot(data=plot_data, x='price_index', y='market_share_pct', 
                size='final_amount_inr_x', hue='brand', sizes=(100, 2000), alpha=0.6)

plt.axvline(1.0, color='red', linestyle='--', label='Market Avg Price')
plt.title(f'Market Intelligence: Brand Positioning in {top_cat}', fontsize=15)
plt.xlabel('Price Index (1.0 = Category Average)')
plt.ylabel('Market Share (%)')
plt.legend(bbox_to_anchor=(1.05, 1))
plt.show()
```

```python
# Pricing Intelligence: Discount Sensitivity
discount_intel = main_df.groupby('brand').agg({
    'discount_percent': 'mean',
    'final_amount_inr': 'sum'
}).reset_index()

plt.figure(figsize=(12, 6))
sns.regplot(data=discount_intel, x='discount_percent', y='final_amount_inr', color='purple')

plt.title('Market Intelligence: Discount Sensitivity Analysis', fontsize=14)
plt.xlabel('Average Discount (%)')
plt.ylabel('Total Brand Revenue (INR)')
plt.show()
```

# Question 29
Build a Seasonal Planning Dashboard for inventory planning, promotional calendar, resource allocation, and seasonal business optimization.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Monthly Revenue Aggregation
seasonal_data = main_df.groupby(main_df['order_date'].dt.month)['final_amount_inr'].mean().reset_index()
seasonal_data.columns = ['month', 'avg_monthly_rev']

# 2. Calculate Seasonality Index (Month Avg / Global Avg)
global_avg = seasonal_data['avg_monthly_rev'].mean()
seasonal_data['seasonality_index'] = seasonal_data['avg_monthly_rev'] / global_avg

# 3. Categorize Seasons (e.g., Summer, Festive, Monsoon)
def get_season_name(m):
    if m in [10, 11]: return 'Festive (Diwali)'
    if m in [3, 4, 5]: return 'Summer Sale'
    if m in [7, 8]: return 'Monsoon/Independence'
    return 'Regular'

seasonal_data['season_type'] = seasonal_data['month'].apply(get_season_name)
```

```python
# Pivot by Month and Category to see "Peak Windows"
pivot_seasonal = main_df.groupby([main_df['order_date'].dt.month, 'category'])['final_amount_inr'].sum().unstack()

plt.figure(figsize=(14, 8))
sns.heatmap(pivot_seasonal, cmap='YlOrRd', annot=False, cbar_kws={'label': 'Revenue Intensity'})

plt.title('Promotional Calendar: Peak Demand Windows by Category', fontsize=15)
plt.ylabel('Month (1=Jan, 12=Dec)')
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.barplot(data=seasonal_data, x='month', y='seasonality_index', hue='season_type', dodge=False)

plt.axhline(1.0, color='black', linestyle='--', label='Yearly Baseline')
plt.title('Inventory Loading Guide: Seasonality Index by Month', fontsize=14)
plt.ylabel('Index (1.0 = Normal)')
plt.legend(title='Season Type')
plt.show()
```

```python
# Monthly average delivery days
logistics_stress = main_df.groupby(main_df['order_date'].dt.month)['delivery_days'].mean()

plt.figure(figsize=(12, 5))
logistics_stress.plot(kind='line', marker='D', color='darkorange', linewidth=2)

plt.title('Logistics Resilience: Seasonal Impact on Delivery Speed', fontsize=14)
plt.ylabel('Average Delivery Days')
plt.xlabel('Month')
plt.grid(True, alpha=0.3)
plt.show()
```

# Question 30
Design a Business Intelligence Command Center integrating all key metrics, automated alerts, performance monitoring, and strategic decision support tools.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Normalize metrics to a 0-100 scale for the "Health Score"
def normalize(series):
    return (series - series.min()) / (series.max() - series.min()) * 100

# 2. Aggregate monthly performance
monthly_hq = main_df.groupby(main_df['order_date'].dt.to_period('M')).agg(
    rev=('final_amount_inr', 'sum'),
    delivery_speed=('delivery_days', 'mean'),
    csat=('customer_rating', 'mean')
).reset_index()

monthly_hq['health_score'] = (
    normalize(monthly_hq['rev']) * 0.4 + 
    (100 - normalize(monthly_hq['delivery_speed'])) * 0.3 + 
    normalize(monthly_hq['csat']) * 0.3
)
```

```python
plt.figure(figsize=(12, 5))
plt.plot(monthly_hq['order_date'].astype(str), monthly_hq['health_score'], 
         marker='o', color='teal', linewidth=3, markersize=10)

plt.axhline(70, color='red', linestyle='--', label='Critical Alert Threshold')
plt.fill_between(monthly_hq['order_date'].astype(str), 70, 100, color='green', alpha=0.1)

plt.title('Business Command Center: Monthly Health Score (2025-2026)', fontsize=15)
plt.ylabel('Health Index (0-100)')
plt.xticks(rotation=45)
plt.grid(axis='y', alpha=0.3)
plt.show()
```
